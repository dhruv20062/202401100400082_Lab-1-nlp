```python
# ============================================================
# AMAZON REVIEWS
# BERT SENTIMENT ANALYSIS + ABSA
# COMPLETE CODE
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q transformers sentencepiece spacy seaborn matplotlib pandas
!python -m spacy download en_core_web_sm -q


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import spacy
import torch

from transformers import pipeline

warnings.filterwarnings("ignore")

print("PyTorch:", torch.__version__)
print("GPU available:", torch.cuda.is_available())

DEVICE = 0 if torch.cuda.is_available() else -1


# ============================================================
# 3. FIND / UPLOAD AMAZON REVIEWS DATASET
# ============================================================

possible_paths = [
    "/content/Amazon_Reviews.csv",
    "/mnt/data/Amazon_Reviews.csv",
    "Amazon_Reviews.csv",
    "./Amazon_Reviews.csv"
]

FILE_PATH = None

for path in possible_paths:
    if os.path.exists(path):
        FILE_PATH = path
        break


# If running on Google Colab and file is not already available
if FILE_PATH is None:

    try:
        from google.colab import files

        print("Upload Amazon_Reviews.csv")

        uploaded = files.upload()

        FILE_PATH = next(iter(uploaded.keys()))

    except ImportError:

        raise FileNotFoundError(
            "Amazon_Reviews.csv not found. "
            "Place it in the same folder as this notebook."
        )


print("Using dataset:", FILE_PATH)


# ============================================================
# 4. LOAD DATASET
# ============================================================

df = pd.read_csv(
    FILE_PATH,
    engine="python",
    on_bad_lines="skip"
)

print("\nDataset loaded successfully")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION")

df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# 6. AUTOMATICALLY DETECT REVIEW COLUMN
# ============================================================

possible_review_columns = [
    "Review Text",
    "review_text",
    "Review",
    "review",
    "Text",
    "text",
    "content",
    "Content",
    "reviews.text"
]

REVIEW_COLUMN = None

for column in possible_review_columns:

    if column in df.columns:

        REVIEW_COLUMN = column
        break


if REVIEW_COLUMN is None:

    text_columns = df.select_dtypes(
        include=["object"]
    ).columns.tolist()

    if len(text_columns) == 0:

        raise ValueError(
            "No text column found in dataset."
        )

    # Choose object column having longest average text
    avg_lengths = {}

    for column in text_columns:

        avg_lengths[column] = (
            df[column]
            .dropna()
            .astype(str)
            .str.len()
            .mean()
        )

    REVIEW_COLUMN = max(
        avg_lengths,
        key=avg_lengths.get
    )


print(
    "\nReview column detected:",
    REVIEW_COLUMN
)


# ============================================================
# 7. AUTOMATICALLY DETECT RATING COLUMN
# ============================================================

possible_rating_columns = [
    "Rating",
    "rating",
    "Score",
    "score",
    "Stars",
    "stars",
    "reviews.rating"
]

RATING_COLUMN = None

for column in possible_rating_columns:

    if column in df.columns:

        RATING_COLUMN = column
        break


print(
    "Rating column detected:",
    RATING_COLUMN
)


# ============================================================
# 8. REMOVE MISSING / EMPTY REVIEWS
# ============================================================

df = df.dropna(
    subset=[REVIEW_COLUMN]
).copy()

df[REVIEW_COLUMN] = (
    df[REVIEW_COLUMN]
    .astype(str)
)

df = df[
    df[REVIEW_COLUMN]
    .str.strip()
    .ne("")
].copy()


# Remove duplicate reviews

df = df.drop_duplicates(
    subset=[REVIEW_COLUMN]
)

df = df.reset_index(
    drop=True
)

print(
    "\nShape after cleaning:",
    df.shape
)


# ============================================================
# 9. CLEAN REVIEW TEXT
# ============================================================

def clean_review(text):

    text = str(text)

    # Remove URLs
    text = re.sub(
        r"http\S+|www\.\S+",
        " ",
        text
    )

    # Remove HTML
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Remove excessive spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


df["Clean_Review"] = (
    df[REVIEW_COLUMN]
    .apply(clean_review)
)


print("\nCleaned Reviews:")

display(
    df[
        [
            REVIEW_COLUMN,
            "Clean_Review"
        ]
    ].head()
)


# ============================================================
# 10. PROCESS RATING COLUMN
# ============================================================

if RATING_COLUMN is not None:

    # Extract first numeric value
    df["Rating_Number"] = (
        df[RATING_COLUMN]
        .astype(str)
        .str.extract(
            r"(\d+(?:\.\d+)?)"
        )[0]
    )

    df["Rating_Number"] = pd.to_numeric(
        df["Rating_Number"],
        errors="coerce"
    )


    print("\nRating Distribution:")

    print(
        df["Rating_Number"]
        .value_counts()
        .sort_index()
    )


# ============================================================
# 11. CREATE RATING SENTIMENT
# ============================================================

def rating_to_sentiment(rating):

    if pd.isna(rating):
        return "UNKNOWN"

    if rating <= 2:
        return "NEGATIVE"

    elif rating == 3:
        return "NEUTRAL"

    else:
        return "POSITIVE"


if "Rating_Number" in df.columns:

    df["Rating_Sentiment"] = (
        df["Rating_Number"]
        .apply(rating_to_sentiment)
    )

    print(
        "\nRating Sentiment Distribution:"
    )

    print(
        df["Rating_Sentiment"]
        .value_counts()
    )


# ============================================================
# 12. VISUALIZE STAR RATINGS
# ============================================================

if "Rating_Number" in df.columns:

    plt.figure(
        figsize=(8, 5)
    )

    sns.countplot(
        data=df,
        x="Rating_Number"
    )

    plt.title(
        "Amazon Review Rating Distribution"
    )

    plt.xlabel(
        "Star Rating"
    )

    plt.ylabel(
        "Number of Reviews"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 13. LOAD BERT SENTIMENT MODEL
# ============================================================

BERT_MODEL = (
    "distilbert-base-uncased-finetuned-sst-2-english"
)

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model=BERT_MODEL,
    tokenizer=BERT_MODEL,
    device=DEVICE
)

print(
    "\nBERT Sentiment Model Loaded"
)


# ============================================================
# 14. TEST BERT
# ============================================================

test_sentences = [
    "The product is excellent.",
    "The product is terrible.",
    "Amazon delivered my order quickly.",
    "Customer service was very disappointing."
]


print("\nBERT TEST")


for sentence in test_sentences:

    prediction = sentiment_pipeline(
        sentence
    )[0]

    print("\nText:", sentence)

    print(
        "Sentiment:",
        prediction["label"]
    )

    print(
        "Confidence:",
        f"{prediction['score']:.2%}"
    )


# ============================================================
# 15. BERT SENTIMENT FUNCTION
# ============================================================

def analyze_sentiment(text):

    text = str(text).strip()

    if not text:

        return "UNKNOWN", 0.0

    prediction = sentiment_pipeline(
        text,
        truncation=True,
        max_length=512
    )[0]

    return (
        prediction["label"],
        prediction["score"]
    )


# ============================================================
# 16. SELECT SAMPLE
# Change 500 to any number you want
# ============================================================

SAMPLE_SIZE = min(
    500,
    len(df)
)

bert_df = df.sample(
    n=SAMPLE_SIZE,
    random_state=42
).copy()

bert_df = bert_df.reset_index(
    drop=True
)

print(
    "\nReviews selected:",
    len(bert_df)
)


# ============================================================
# 17. RUN BERT ON DATASET
# ============================================================

bert_labels = []
bert_confidences = []


for i, review in enumerate(
    bert_df["Clean_Review"]
):

    try:

        label, confidence = (
            analyze_sentiment(
                review
            )
        )

    except Exception as e:

        label = "ERROR"
        confidence = 0.0


    bert_labels.append(
        label
    )

    bert_confidences.append(
        confidence
    )


    if (i + 1) % 50 == 0:

        print(
            f"Processed "
            f"{i + 1}/"
            f"{len(bert_df)}"
        )


bert_df[
    "BERT_Sentiment"
] = bert_labels


bert_df[
    "BERT_Confidence"
] = bert_confidences


bert_df[
    "BERT_Confidence_Percent"
] = (
    bert_df[
        "BERT_Confidence"
    ] * 100
).round(2)


# ============================================================
# 18. DISPLAY BERT RESULTS
# ============================================================

display_columns = [
    REVIEW_COLUMN
]

if "Rating_Number" in bert_df.columns:

    display_columns.append(
        "Rating_Number"
    )

if "Rating_Sentiment" in bert_df.columns:

    display_columns.append(
        "Rating_Sentiment"
    )

display_columns.extend(
    [
        "BERT_Sentiment",
        "BERT_Confidence_Percent"
    ]
)


display(
    bert_df[
        display_columns
    ].head(20)
)


# ============================================================
# 19. BERT SENTIMENT DISTRIBUTION
# ============================================================

print(
    "\nBERT Sentiment Distribution:"
)

print(
    bert_df[
        "BERT_Sentiment"
    ].value_counts()
)


# ============================================================
# 20. VISUALIZE BERT SENTIMENT
# ============================================================

plt.figure(
    figsize=(8, 5)
)

sns.countplot(
    data=bert_df,
    x="BERT_Sentiment"
)

plt.title(
    "BERT Sentiment Distribution"
)

plt.xlabel(
    "Sentiment"
)

plt.ylabel(
    "Number of Reviews"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. BERT CONFIDENCE DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10, 5)
)

sns.histplot(
    data=bert_df,
    x="BERT_Confidence",
    hue="BERT_Sentiment",
    bins=20
)

plt.title(
    "BERT Confidence Distribution"
)

plt.xlabel(
    "Confidence"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. COMPARE RATING SENTIMENT WITH BERT
# ============================================================

if "Rating_Sentiment" in bert_df.columns:

    comparison_table = pd.crosstab(
        bert_df[
            "Rating_Sentiment"
        ],
        bert_df[
            "BERT_Sentiment"
        ]
    )

    print(
        "\nRating vs BERT Sentiment:"
    )

    display(
        comparison_table
    )


# ============================================================
# 23. VISUALIZE RATING VS BERT
# ============================================================

if "Rating_Number" in bert_df.columns:

    plt.figure(
        figsize=(9, 5)
    )

    sns.countplot(
        data=bert_df,
        x="Rating_Number",
        hue="BERT_Sentiment"
    )

    plt.title(
        "Star Rating vs BERT Sentiment"
    )

    plt.xlabel(
        "Star Rating"
    )

    plt.ylabel(
        "Number of Reviews"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 24. CALCULATE BERT/RATING AGREEMENT
# ============================================================

agreement_percentage = None


if "Rating_Sentiment" in bert_df.columns:

    comparison_sample = bert_df[
        bert_df[
            "Rating_Sentiment"
        ].isin(
            [
                "POSITIVE",
                "NEGATIVE"
            ]
        )
    ].copy()


    comparison_sample[
        "Agreement"
    ] = (
        comparison_sample[
            "Rating_Sentiment"
        ]
        ==
        comparison_sample[
            "BERT_Sentiment"
        ]
    )


    agreement_percentage = (
        comparison_sample[
            "Agreement"
        ].mean()
        * 100
    )


    print(
        "\nBERT vs Rating Agreement:",
        round(
            agreement_percentage,
            2
        ),
        "%"
    )


# ============================================================
# 25. SHOW DISAGREEMENTS
# ============================================================

if (
    "Rating_Sentiment" in bert_df.columns
    and agreement_percentage is not None
):

    disagreements = (
        comparison_sample[
            comparison_sample[
                "Agreement"
            ] == False
        ]
    )


    print(
        "\nNumber of Disagreements:",
        len(disagreements)
    )


    disagreement_columns = [
        REVIEW_COLUMN,
        "Rating_Number",
        "Rating_Sentiment",
        "BERT_Sentiment",
        "BERT_Confidence_Percent"
    ]


    display(
        disagreements[
            disagreement_columns
        ].head(10)
    )


# ============================================================
# 26. LOAD SPACY
# ============================================================

nlp = spacy.load(
    "en_core_web_sm"
)

print(
    "\nspaCy loaded"
)


# ============================================================
# 27. LOAD ABSA MODEL
# ============================================================

ABSA_MODEL = (
    "yangheng/deberta-v3-base-absa-v1.1"
)

absa_pipeline = pipeline(
    "text-classification",
    model=ABSA_MODEL,
    tokenizer=ABSA_MODEL,
    device=DEVICE
)

print(
    "\nABSA Model Loaded"
)


# ============================================================
# 28. ABSA FUNCTION
# ============================================================

def analyze_aspects(
    sentence,
    aspects
):

    results = []

    sentence = str(
        sentence
    )


    for aspect in aspects:

        try:

            prediction = (
                absa_pipeline(
                    sentence,
                    text_pair=aspect,
                    truncation=True,
                    max_length=512
                )[0]
            )


            results.append(
                {
                    "Aspect":
                        aspect,

                    "Sentiment":
                        prediction[
                            "label"
                        ],

                    "Confidence":
                        prediction[
                            "score"
                        ]
                }
            )


        except Exception:

            results.append(
                {
                    "Aspect":
                        aspect,

                    "Sentiment":
                        "ERROR",

                    "Confidence":
                        0.0
                }
            )


    return pd.DataFrame(
        results
    )


# ============================================================
# 29. AMAZON ASPECTS
# ============================================================

amazon_aspects = [
    "product",
    "quality",
    "price",
    "delivery",
    "shipping",
    "packaging",
    "customer service",
    "service",
    "refund",
    "return",
    "seller",
    "payment",
    "order",
    "driver",
    "support",
    "replacement"
]


# ============================================================
# 30. FIND AMAZON ASPECTS IN REVIEW
# ============================================================

def find_amazon_aspects(
    text
):

    text = str(
        text
    ).lower()


    found = []


    for aspect in amazon_aspects:

        if aspect.lower() in text:

            found.append(
                aspect
            )


    return found


# ============================================================
# 31. DETECT ASPECTS FOR DATASET
# ============================================================

bert_df[
    "Detected_Aspects"
] = (
    bert_df[
        "Clean_Review"
    ]
    .apply(
        find_amazon_aspects
    )
)


display(
    bert_df[
        [
            REVIEW_COLUMN,
            "Detected_Aspects"
        ]
    ].head(20)
)


# ============================================================
# 32. AUTOMATIC NOUN-PHRASE ASPECT EXTRACTION
# ============================================================

def extract_noun_aspects(
    text
):

    text = str(
        text
    )[:5000]


    doc = nlp(
        text
    )


    aspects = []

    seen = set()


    for chunk in doc.noun_chunks:

        aspect = (
            chunk.text
            .strip()
            .lower()
        )


        aspect = re.sub(
            r"^(the|a|an|my|this|that|their|our|your|its)\s+",
            "",
            aspect
        )


        if (
            len(aspect) > 2
            and
            len(
                aspect.split()
            ) <= 4
            and
            aspect not in seen
        ):

            aspects.append(
                aspect
            )

            seen.add(
                aspect
            )


    return aspects[:10]


# ============================================================
# 33. TEST ASPECT EXTRACTION
# ============================================================

example_review = (
    bert_df[
        "Clean_Review"
    ].iloc[0]
)


print(
    "\nExample Review:"
)

print(
    example_review
)


automatic_aspects = (
    extract_noun_aspects(
        example_review
    )
)


print(
    "\nAutomatically Extracted Aspects:"
)

print(
    automatic_aspects
)


# ============================================================
# 34. AUTOMATIC ABSA
# ============================================================

def automatic_absa(
    sentence,
    max_aspects=8
):

    aspects = (
        find_amazon_aspects(
            sentence
        )
    )


    # If predefined Amazon aspects are not found,
    # use automatic noun extraction

    if len(aspects) == 0:

        aspects = (
            extract_noun_aspects(
                sentence
            )
        )


    aspects = aspects[
        :max_aspects
    ]


    if len(aspects) == 0:

        return pd.DataFrame(
            columns=[
                "Aspect",
                "Sentiment",
                "Confidence"
            ]
        )


    return analyze_aspects(
        sentence,
        aspects
    )


# ============================================================
# 35. TEST AUTOMATIC ABSA
# ============================================================

auto_result = (
    automatic_absa(
        example_review
    )
)


if not auto_result.empty:

    auto_result[
        "Confidence_Percent"
    ] = (
        auto_result[
            "Confidence"
        ] * 100
    ).round(2)


    print(
        "\nAutomatic ABSA Result:"
    )


    display(
        auto_result[
            [
                "Aspect",
                "Sentiment",
                "Confidence_Percent"
            ]
        ]
    )


# ============================================================
# 36. RUN ABSA ON DATASET
#
# ABSA is slower than normal BERT sentiment.
# Increase 100 if you want more reviews.
# ============================================================

ABSA_SAMPLE_SIZE = min(
    100,
    len(bert_df)
)


absa_sample = (
    bert_df
    .head(
        ABSA_SAMPLE_SIZE
    )
    .copy()
)


all_absa_results = []


for position, row in (
    absa_sample.iterrows()
):


    review = row[
        "Clean_Review"
    ]


    aspects = (
        find_amazon_aspects(
            review
        )
    )


    if len(aspects) == 0:

        aspects = (
            extract_noun_aspects(
                review
            )[:5]
        )


    if len(aspects) == 0:

        continue


    results = (
        analyze_aspects(
            review,
            aspects
        )
    )


    for _, result in (
        results.iterrows()
    ):


        record = {

            "Review_Index":
                position,

            "Review":
                review,

            "Aspect":
                result[
                    "Aspect"
                ],

            "Aspect_Sentiment":
                result[
                    "Sentiment"
                ],

            "Confidence":
                result[
                    "Confidence"
                ]
        }


        if "Rating_Number" in row.index:

            record[
                "Rating"
            ] = row[
                "Rating_Number"
            ]


        all_absa_results.append(
            record
        )


    if (
        position + 1
    ) % 10 == 0:

        print(
            f"ABSA processed "
            f"{position + 1}/"
            f"{ABSA_SAMPLE_SIZE}"
        )


amazon_absa_df = pd.DataFrame(
    all_absa_results
)


# ============================================================
# 37. DISPLAY ABSA RESULTS
# ============================================================

if not amazon_absa_df.empty:

    amazon_absa_df[
        "Confidence_Percent"
    ] = (
        amazon_absa_df[
            "Confidence"
        ] * 100
    ).round(2)


    print(
        "\nABSA Results:"
    )


    display(
        amazon_absa_df.head(
            30
        )
    )


# ============================================================
# 38. ASPECT FREQUENCY
# ============================================================

if not amazon_absa_df.empty:

    aspect_frequency = (
        amazon_absa_df[
            "Aspect"
        ]
        .value_counts()
        .reset_index()
    )


    aspect_frequency.columns = [
        "Aspect",
        "Count"
    ]


    print(
        "\nAspect Frequency:"
    )


    display(
        aspect_frequency
    )


# ============================================================
# 39. VISUALIZE MOST DISCUSSED ASPECTS
# ============================================================

if not amazon_absa_df.empty:

    top_aspects = (
        amazon_absa_df[
            "Aspect"
        ]
        .value_counts()
        .head(15)
        .index
    )


    plot_df = (
        amazon_absa_df[
            amazon_absa_df[
                "Aspect"
            ].isin(
                top_aspects
            )
        ]
    )


    plt.figure(
        figsize=(10, 7)
    )


    sns.countplot(
        data=plot_df,
        y="Aspect",
        order=top_aspects
    )


    plt.title(
        "Most Discussed Aspects"
    )

    plt.xlabel(
        "Number of Mentions"
    )

    plt.ylabel(
        "Aspect"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 40. ASPECT SENTIMENT TABLE
# ============================================================

if not amazon_absa_df.empty:

    aspect_sentiment = (
        pd.crosstab(
            amazon_absa_df[
                "Aspect"
            ],
            amazon_absa_df[
                "Aspect_Sentiment"
            ]
        )
    )


    print(
        "\nAspect Sentiment Distribution:"
    )


    display(
        aspect_sentiment
    )


# ============================================================
# 41. VISUALIZE ASPECT SENTIMENT
# ============================================================

if not amazon_absa_df.empty:

    top_aspects = (
        amazon_absa_df[
            "Aspect"
        ]
        .value_counts()
        .head(15)
        .index
    )


    plot_df = (
        amazon_absa_df[
            amazon_absa_df[
                "Aspect"
            ].isin(
                top_aspects
            )
        ]
    )


    plt.figure(
        figsize=(12, 7)
    )


    sns.countplot(
        data=plot_df,
        y="Aspect",
        hue="Aspect_Sentiment",
        order=top_aspects
    )


    plt.title(
        "Aspect-Based Sentiment Analysis"
    )

    plt.xlabel(
        "Number of Reviews"
    )

    plt.ylabel(
        "Aspect"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 42. AVERAGE ABSA CONFIDENCE
# ============================================================

if not amazon_absa_df.empty:

    average_confidence = (
        amazon_absa_df
        .groupby(
            [
                "Aspect",
                "Aspect_Sentiment"
            ]
        )[
            "Confidence"
        ]
        .mean()
        .reset_index()
    )


    average_confidence[
        "Confidence_Percent"
    ] = (
        average_confidence[
            "Confidence"
        ] * 100
    ).round(2)


    print(
        "\nAverage ABSA Confidence:"
    )


    display(
        average_confidence
        .sort_values(
            "Confidence",
            ascending=False
        )
    )


# ============================================================
# 43. REVIEW LENGTH
# ============================================================

bert_df[
    "Review_Length"
] = (
    bert_df[
        "Clean_Review"
    ]
    .str.split()
    .str.len()
)


print(
    "\nReview Length:"
)


display(
    bert_df[
        [
            REVIEW_COLUMN,
            "Review_Length",
            "BERT_Sentiment",
            "BERT_Confidence_Percent"
        ]
    ].head()
)


# ============================================================
# 44. REVIEW LENGTH VISUALIZATION
# ============================================================

plt.figure(
    figsize=(10, 5)
)


sns.histplot(
    data=bert_df,
    x="Review_Length",
    hue="BERT_Sentiment",
    bins=30
)


plt.title(
    "Review Length by Sentiment"
)

plt.xlabel(
    "Number of Words"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ============================================================
# 45. MOST CONFIDENT POSITIVE REVIEWS
# ============================================================

positive_reviews = (
    bert_df[
        bert_df[
            "BERT_Sentiment"
        ] == "POSITIVE"
    ]
    .sort_values(
        "BERT_Confidence",
        ascending=False
    )
)


print(
    "\nMost Confident Positive Reviews:"
)


positive_columns = [
    REVIEW_COLUMN,
    "BERT_Confidence_Percent"
]


if "Rating_Number" in positive_reviews.columns:

    positive_columns.insert(
        1,
        "Rating_Number"
    )


display(
    positive_reviews[
        positive_columns
    ].head(10)
)


# ============================================================
# 46. MOST CONFIDENT NEGATIVE REVIEWS
# ============================================================

negative_reviews = (
    bert_df[
        bert_df[
            "BERT_Sentiment"
        ] == "NEGATIVE"
    ]
    .sort_values(
        "BERT_Confidence",
        ascending=False
    )
)


print(
    "\nMost Confident Negative Reviews:"
)


negative_columns = [
    REVIEW_COLUMN,
    "BERT_Confidence_Percent"
]


if "Rating_Number" in negative_reviews.columns:

    negative_columns.insert(
        1,
        "Rating_Number"
    )


display(
    negative_reviews[
        negative_columns
    ].head(10)
)


# ============================================================
# 47. FINAL SUMMARY
# ============================================================

print(
    "\n========================================"
)

print(
    "AMAZON REVIEWS ANALYSIS SUMMARY"
)

print(
    "========================================"
)


print(
    "\nTotal Reviews:",
    len(df)
)


print(
    "Reviews Analyzed with BERT:",
    len(bert_df)
)


print(
    "\nBERT Sentiment Counts:"
)


print(
    bert_df[
        "BERT_Sentiment"
    ].value_counts()
)


print(
    "\nAverage BERT Confidence:"
)


print(
    round(
        bert_df[
            "BERT_Confidence_Percent"
        ].mean(),
        2
    ),
    "%"
)


if agreement_percentage is not None:

    print(
        "\nBERT vs Rating Agreement:"
    )

    print(
        round(
            agreement_percentage,
            2
        ),
        "%"
    )


if not amazon_absa_df.empty:

    print(
        "\nMost Discussed Aspects:"
    )


    print(
        amazon_absa_df[
            "Aspect"
        ]
        .value_counts()
        .head(10)
    )


    print(
        "\nAspect Sentiment Counts:"
    )


    print(
        amazon_absa_df[
            "Aspect_Sentiment"
        ]
        .value_counts()
    )


# ============================================================
# 48. SAVE BERT RESULTS
# ============================================================

bert_output = (
    "Amazon_BERT_Sentiment_Results.csv"
)


bert_df.to_csv(
    bert_output,
    index=False
)


print(
    "\nSaved:",
    bert_output
)


# ============================================================
# 49. SAVE ABSA RESULTS
# ============================================================

if not amazon_absa_df.empty:

    absa_output = (
        "Amazon_ABSA_Results.csv"
    )


    amazon_absa_df.to_csv(
        absa_output,
        index=False
    )


    print(
        "Saved:",
        absa_output
    )


# ============================================================
# 50. INTERACTIVE REVIEW ANALYSIS
# ============================================================

print(
    "\n========================================"
)

print(
    "INTERACTIVE SENTIMENT + ABSA"
)

print(
    "========================================"
)


user_review = input(
    "\nEnter an Amazon review: "
)


# BERT prediction

bert_prediction = (
    sentiment_pipeline(
        user_review,
        truncation=True,
        max_length=512
    )[0]
)


print(
    "\nBERT SENTIMENT"
)


print(
    "Sentiment:",
    bert_prediction[
        "label"
    ]
)


print(
    "Confidence:",
    f"{bert_prediction['score']:.2%}"
)


# ============================================================
# 51. DETECT ASPECTS FROM USER REVIEW
# ============================================================

detected_aspects = (
    find_amazon_aspects(
        user_review
    )
)


if len(
    detected_aspects
) == 0:

    detected_aspects = (
        extract_noun_aspects(
            user_review
        )[:8]
    )


print(
    "\nDetected Aspects:"
)

print(
    detected_aspects
)


# ============================================================
# 52. INTERACTIVE ABSA
# ============================================================

if detected_aspects:

    interactive_absa = (
        analyze_aspects(
            user_review,
            detected_aspects
        )
    )


    interactive_absa[
        "Confidence_Percent"
    ] = (
        interactive_absa[
            "Confidence"
        ] * 100
    ).round(2)


    print(
        "\nASPECT-BASED SENTIMENT"
    )


    display(
        interactive_absa[
            [
                "Aspect",
                "Sentiment",
                "Confidence_Percent"
            ]
        ]
    )


else:

    print(
        "\nNo aspects detected."
    )
```
